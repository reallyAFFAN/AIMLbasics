```python
import heapq

def uniform_cost_search(graph, start, goal):

    # Priority queue to store (cost, node, path)
    priority_queue = [(0, start, [start])]
    # Set to keep track of visited nodes
    visited = set()

    while priority_queue:
        cost, current_node, path = heapq.heappop(priority_queue)

        if current_node in visited:
            continue

        visited.add(current_node)

        if current_node == goal:
            return path, cost

        for neighbor, edge_cost in graph.get(current_node, {}).items():
            if neighbor not in visited:
                heapq.heappush(priority_queue, (cost + edge_cost, neighbor, path + [neighbor]))

    return None, None

# Example Usage for UCS
print("--- Uniform Cost Search Example ---")
ucs_graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 2, 'E': 5},
    'C': {'F': 1},
    'D': {'G': 3},
    'E': {'G': 1},
    'F': {'G': 2}
}
start_node_ucs = 'A'
goal_node_ucs = 'G'

path_ucs, cost_ucs = uniform_cost_search(ucs_graph, start_node_ucs, goal_node_ucs)

if path_ucs:
    print(f"Path from {start_node_ucs} to {goal_node_ucs}: {path_ucs}")
    print(f"Total cost: {cost_ucs}")
else:
    print(f"No path found from {start_node_ucs} to {goal_node_ucs}")
```

```python
import heapq

def a_star_search(graph, start, goal, heuristic):

    # Priority queue to store (f_cost, g_cost, node, path)
    # f_cost = g_cost + h_cost
    priority_queue = [(heuristic[start], 0, start, [start])]
    # Dictionary to store the minimum cost to reach each node found so far
    g_costs = {node: float('inf') for node in graph}
    g_costs[start] = 0

    while priority_queue:
        f_cost, g_cost, current_node, path = heapq.heappop(priority_queue)

        # If we have found a shorter path to current_node already, skip
        if g_cost > g_costs[current_node]:
            continue

        if current_node == goal:
            return path, g_cost

        for neighbor, edge_cost in graph.get(current_node, {}).items():
            tentative_g_cost = g_cost + edge_cost

            if tentative_g_cost < g_costs[neighbor]:
                g_costs[neighbor] = tentative_g_cost
                f_cost = tentative_g_cost + heuristic.get(neighbor, 0) # Use 0 if heuristic not defined for neighbor
                heapq.heappush(priority_queue, (f_cost, tentative_g_cost, neighbor, path + [neighbor]))

    return None, None

# Example Usage for A*
print("\n--- A* Algorithm Example ---")
a_star_graph = {
    'A': {'B': 6, 'D': 3},
    'B': {'E': 2},
    'D': {'B': 2, 'E': 4},
    'E': {'C': 1},
    'C': {}
}

a_star_heuristic = {
    'A': 8,
    'B': 4,
    'C': 0,
    'D': 2,
    'E': 1
}
start_node_a_star = 'A'
goal_node_a_star = 'C'

path_a_star, cost_a_star = a_star_search(a_star_graph, start_node_a_star, goal_node_a_star, a_star_heuristic)

if path_a_star:
    print(f"Path from {start_node_a_star} to {goal_node_a_star}: {path_a_star}")
    print(f"Total cost: {cost_a_star}")
else:
    print(f"No path found from {start_node_a_star} to {goal_node_a_star}")
```
